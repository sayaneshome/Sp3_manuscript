```python
from Bio import motifs, SeqIO
from Bio.Seq import Seq

try:

    jaspar_path = "jaspa.jaspar"
    with open(jaspar_path) as handle:
        motif_list = motifs.parse(handle, "jaspar")
    print(f"Loaded {len(motif_list)} motifs.")


    threshold = 8  
    window_size = 10  

    # File paths
    fasta_file_path = "upstream_sequences_LMC_GC.fasta"
    output_file_path = "motif_scan_results_LMC_GC_DE_v1_w10_t8.txt"

    with open(output_file_path, 'w') as output_file:
        for record in SeqIO.parse(fasta_file_path, "fasta"):
            sequence = record.seq
            gene_name = record.id

            for motif in motif_list:
                for position in range(len(sequence) - window_size + 1):
                    window_seq = sequence[position:position + window_size]
                    total_score = 0
                    valid_scores = 0

                    for i in range(len(window_seq) - len(motif) + 1):
                        subseq = window_seq[i:i + len(motif)]
                        score = motif.pssm.calculate(subseq)
                        if score != float('-inf'):
                            total_score += score
                            valid_scores += 1

                    if valid_scores > 0:
                        average_score = total_score / valid_scores
                        if average_score > threshold:
                            output_file.write(f"Gene: {gene_name}, Motif: {motif.name}, Start Position: {position}, End Position: {position + window_size}, Average Score: {average_score}\n")

    print("Motif scanning complete. Results saved to", output_file_path)

except FileNotFoundError as e:
    print(f"File not found: {e}")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
def extract_gene_names(input_file_path, output_file_path):
    gene_names = set()

    with open(input_file_path, 'r') as file:
        for line in file:
            if line.startswith('Gene:'):
                gene_name = line.split(',')[0].split(':')[1].strip()
                gene_names.add(gene_name)

    with open(output_file_path, 'w') as output_file:
        for gene_name in sorted(gene_names):
            output_file.write(gene_name + '\n')

    return len(gene_names)

# File paths
input_file_path = 'motif_scan_results_LMC_GC_DE_v1_w10_t8_tbp_excluded.txt'
output_file_path = 'unique_gene_names.txt'

# Extract gene names and report the number
num_genes = extract_gene_names(input_file_path, output_file_path)
print(f'Number of unique genes: {num_genes}')
```

```python
from Bio import SeqIO

def count_fasta_entries(fasta_file_path):
    count = 0
    with open(fasta_file_path, "r") as fasta_file:
        for record in SeqIO.parse(fasta_file, "fasta"):
            count += 1
    return count

# File path
fasta_file_path = "upstream_sequences_LMC_GC.fasta"

# Count entries in the FASTA file
num_entries = count_fasta_entries(fasta_file_path)
print(f"Number of entries in the file: {num_entries}")
```
