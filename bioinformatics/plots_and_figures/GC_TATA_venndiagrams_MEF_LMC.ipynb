```r
library(VennDiagram)
library(dplyr)
```

```r
csv1 <- read.csv('outlier_genecard_genes_with_both_tata_and_gc_box_LMC_total_Dec5_2023.csv')
csv2 <- read.csv('outlier_genecard_genes_with_gc_box_LMC_total_Dec5_2023.csv')
csv3 <- read.csv('outlier_genecard_genes_with_tata_box_LMC_total_Dec5_2023.csv')
```

```r
csv4 <- read.csv('TATA_LMC_Dec6_2023.csv')
csv5 <- read.csv('GCbox_LMC_Dec6_2023.csv')
csv6 <- read.csv('otherTFs_LMC_Dec6_2023.csv')
```

```r
csv4$X <- NULL
csv5$X <- NULL
csv6$X <- NULL
```

```r
colnames(csv1)[1] <- "Gene_name"
colnames(csv2)[1] <- "Gene_name"
colnames(csv3)[1] <- "Gene_name"
colnames(csv4)[1] <- "Gene_name"
colnames(csv5)[1] <- "Gene_name"
colnames(csv6)[1] <- "Gene_name"
```

```r
csv4_5 <- merge(csv4,csv5,by = "Gene_name")
```

```r
csv_tata_lmc <- unique(rbind(csv3,csv4))
csv_gc_lmc <- unique(rbind(csv2,csv5))
csv_tata_gc_lmc <- unique(rbind(csv1,csv4_5))
```

```r
nrow(csv_tata_lmc)
nrow(csv_gc_lmc)
nrow(csv_tata_gc_lmc)
```

```r
gene_lists <- list(csv_tata_lmc$Gene_name, csv_gc_lmc$Gene_name)

# Run the venn diagram function
venn.plot <- venn.diagram(
    x = gene_lists,
    filename = NULL,
    category.names = c("", ""),  # Set as empty strings
    output = TRUE,
    fill = c("red", "blue"),
    alpha = 0.5,
    fontface = "bold",
    fontsize = 20,
    cat.fontface = "bold",
    cat.fontsize = 20,
    lty = "solid",
    lwd = 2,
    label.col = "black",  # Set this to "black" to show numbers
    category.pos = NULL  # Set to NULL to avoid positioning non-existing labels
)

# Plot the diagram
grid.draw(venn.plot)
```
